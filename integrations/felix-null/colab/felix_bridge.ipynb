# Felix.Null Colab Bridge
Compute limb for the Felix.Null working lattice.

This notebook mounts Google Drive, syncs the Paramecia core repository, validates Felix's configured Drive anchors, and writes a non-destructive runtime health artifact.


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from pathlib import Path
import json, subprocess, datetime, os

REPO_URL = 'https://github.com/simplyabbeyd-pixel/paramecia-core.git'
REPO_DIR = Path('/content/paramecia-core')

if REPO_DIR.exists():
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL, str(REPO_DIR)], check=True)

print('Repository ready:', REPO_DIR)
```

```python
FELIX = {
    'name': 'Felix.Null',
    'version': '0.1.0',
    'principle': 'Nothing flourishes alone.',
    'drive_ids': {
        'felix_zone': '1LO3aiYRbEA6qUXDQ6D8DG8tzQyAyw1Xp',
        'build_lab': '1wxGmFrEQV8rUmK_jfKStRuyNAM_QAlnR',
        'colab_runtime': '1vrv3ElM_2hfDrX0mKSww1QEnBgfRueJb',
        'apps_script_automation': '1mqD7vvtsc7PyuynTJ3_SiFRz4baDhoU5',
    },
}
FELIX
```

```python
# Colab's mounted Drive is path-oriented, while folder IDs are provider identifiers.
# Keep IDs as canonical anchors and write runtime artifacts to a user-visible Drive path.
RUNTIME_DIR = Path('/content/drive/MyDrive/Felix.Null/Colab Runtime')
RUNTIME_DIR.mkdir(parents=True, exist_ok=True)

health = {
    'ok': True,
    'name': FELIX['name'],
    'version': FELIX['version'],
    'timestamp': datetime.datetime.now(datetime.timezone.utc).isoformat(),
    'repo': str(REPO_DIR),
    'runtime_dir': str(RUNTIME_DIR),
    'drive_ids': FELIX['drive_ids'],
}

health_path = RUNTIME_DIR / 'felix_runtime_health.json'
health_path.write_text(json.dumps(health, indent=2), encoding='utf-8')
print(json.dumps(health, indent=2))
print('Wrote:', health_path)
```
